# Calculadora de Transformações Lineares (Y = aX + b)

Esta ferramenta calcula a Média, Variância e Desvio Padrão de uma nova variável aleatória $Y$ que é uma transformação linear de $X$.

### Propriedades Matemáticas:
- **Média**: $E[aX + b] = a \cdot E[X] + b$
- **Variância**: $Var[aX + b] = a^2 \cdot Var[X]$
- **Desvio Padrão**: $DP[aX + b] = |a| \cdot DP[X]$

---

## ⚙️ Setup

```python
import numpy as np
```

## 🛠️ Função de Cálculo

```python
def calcular_transformacao_linear(mu_x, var_x, a, b):
    """
    Calcula as propriedades de Y = aX + b
    """
    # 1. Média de Y
    mu_y = a * mu_x + b
    
    # 2. Variância de Y
    var_y = (a**2) * var_x
    
    # 3. Desvio Padrão de Y
    std_y = np.sqrt(var_y)
    
    print(f"📊 RESULTADOS PARA Y = {a}X + {b}")
    print(f"{"-"*30}")
    print(f"X original:  E[X] = {mu_x:.4f} | Var[X] = {var_x:.4f} | DP[X] = {np.sqrt(var_x):.4f}")
    print(f"Y resultante: E[Y] = {mu_y:.4f} | Var[Y] = {var_y:.4f} | DP[Y] = {std_y:.4f}")
    print(f"{"-"*30}")
```

## 🚀 Execução

```python
# Altere os valores abaixo e execute (Shift + Enter)
calcular_transformacao_linear(
    mu_x  = 4.95,     # Valor Esperado (Média) de X
    var_x = 0.031**2, # Variância de X (se tiver o DP, eleve ao quadrado)
    a     = 250,      # Coeficiente angular (multiplicador)
    b     = 0         # Constante aditiva (deslocamento)
)
```
