# Calculadora de Transformações Lineares ($X = a + bY$)

Esta ferramenta calcula o Valor Esperado e a Variância de uma nova variável aleatória $X$ que é uma transformação linear de $Y$.

### Fórmulas (conforme material da FEUP):
Considere $Y$ a variável original e $X$ a transformada:
$$X = a + b \cdot Y$$

- **Valor Esperado**: $E(X) = a + b \cdot E(Y)$
- **Variância**: $Var(X) = b^2 \cdot Var(Y)$
- **Desvio Padrão**: $DP(X) = |b| \cdot DP(Y)$

---

## ⚙️ Setup

```python
import numpy as np
print("✅ Setup concluído!")
```

## 🛠️ Função de Cálculo

```python
def calcular_transformacao_linear(mu_y, var_y, a, b):
    """
    Calcula as propriedades de X = a + b*Y
    Parametros:
    mu_y: E(Y) da variavel original
    var_y: Var(Y) da variavel original
    a: constante aditiva
    b: coeficiente multiplicativo
    """
    # 1. Valor Esperado de X
    mu_x = a + b * mu_y
    
    # 2. Variância de X
    var_x = (b**2) * var_y
    
    # 3. Desvio Padrão de X
    std_x = np.sqrt(var_x)
    
    print(f"📊 RESULTADOS PARA X = {a} + {b}Y")
    print(f"{'='*40}")
    print(f"Y Original:    E(Y) = {mu_y:.4f}  | Var(Y) = {var_y:.4f} | DP(Y) = {np.sqrt(var_y):.4f}")
    print(f"X Transformada: E(X) = {mu_x:.4f}  | Var(X) = {var_x:.4f} | DP(X) = {std_x:.4f}")
    print(f"{'='*40}")
```

## 🚀 Execução

```python
# Altere os valores abaixo conforme o enunciado e execute (Shift + Enter)
#X = a + b*Y
calcular_transformacao_linear(
    mu_y  = 3.5,     # E(Y) - Média da variável original
    var_y = 1.1, # Var(Y) - Variância da variável original
    a     = 0,        # a - Constante aditiva (deslocamento)
    b     = 7*24       # b - Multiplicador (coeficiente)
)
```

```python
import numpy as np

def calcular_propriedades_estatisticas(mu_x, var_x, n, b=0, tipo="soma_independente"):
    """
    Calcula propriedades de uma nova variável Y baseada em X.
    
    Parâmetros:
    mu_x, var_x: Média e Variância de X
    n: Multiplicador ou número de repetições
    b: Constante aditiva (opcional)
    tipo: 
      - 'soma_independente': Para somar n eventos independentes (ex: horas numa semana).
                             Var(Y) = n * Var(X)
      - 'transformacao_linear': Para escalar uma variável (ex: mudança de unidade).
                                Var(Y) = n^2 * Var(X)
    """
    
    # 1. A Média é igual nos dois casos
    mu_y = n * mu_x + b
    
    # 2. A Variância depende do tipo
    if tipo == "soma_independente":
        # Soma de variâncias: Var(X1 + X2 + ... + Xn) = n * Var(X)
        var_y = n * var_x 
        descricao = f"SOMA de {n} variáveis independentes"
        
    elif tipo == "transformacao_linear":
        # Propriedade da variância: Var(n*X) = n^2 * Var(X)
        var_y = (n**2) * var_x
        descricao = f"TRANSFORMAÇÃO LINEAR Y = {n}X + {b}"
        
    else:
        raise ValueError("Tipo inválido. Use 'soma_independente' ou 'transformacao_linear'")
    
    std_y = np.sqrt(var_y)
    
    print(f"📊 RESULTADOS PARA: {descricao}")
    print(f"{'-'*45}")
    print(f"X original:   E[X] = {mu_x:.4f} | Var[X] = {var_x:.4f}")
    print(f"Y resultante: E[Y] = {mu_y:.4f} | Var[Y] = {var_y:.4f} | DP[Y] = {std_y:.4f}")
    print(f"{'-'*45}\n")
    
    return mu_y, var_y

# --- EXEMPLO DE USO PARA O TEU PROBLEMA ---

# 1. Para calcular o CONSUMO SEMANAL (C)
# Usamos 'soma_independente' porque cada hora varia independentemente
print(">>> CÁLCULO DO CONSUMO SEMANAL (C):")
mu_C, var_C = calcular_propriedades_estatisticas(
    mu_x=3.5, var_x=1.1, n=168, tipo="soma_independente"
)

# 2. Para calcular o EXPOENTE (W = 0.02C - 11.76)
# Usamos 'transformacao_linear' porque W depende diretamente do valor fixo de C
print(">>> CÁLCULO DO EXPOENTE (W):")
# Nota: Aqui a entrada é o C que acabámos de calcular
mu_W, var_W = calcular_propriedades_estatisticas(
    mu_x=mu_C, var_x=var_C, n=0.02, b=-(0.02*588), tipo="transformacao_linear"
)
```
